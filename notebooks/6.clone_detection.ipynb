```python
from sentence_transformers import SentenceTransformer, util

# -------------------------
# Two pieces of code to compare
# -------------------------
code_1 = """
def add(a, b):
    return a + b
"""

code_2 = """
def sum_numbers(x, y):
    result = x + y
    return result
"""

# -------------------------
# Load CodeBERT model
# -------------------------
model = SentenceTransformer("microsoft/codebert-base")

# -------------------------
# Encode both pieces of code
# -------------------------
emb_1 = model.encode(code_1, convert_to_tensor=True, normalize_embeddings=True)
emb_2 = model.encode(code_2, convert_to_tensor=True, normalize_embeddings=True)

# -------------------------
# Compute cosine similarity
# -------------------------
similarity = util.cos_sim(emb_1, emb_2).item()
print(f"Embedding similarity between the two code snippets: {similarity:.4f}")
```
